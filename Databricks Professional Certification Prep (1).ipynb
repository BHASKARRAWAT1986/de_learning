### 1. Delta Sharing

**Key Points:**
* Open protocol — REST API + short-lived pre-signed URLs (cloud-agnostic)
* Two recipient types: **Open** (token-based, any platform) and **Databricks-to-Databricks** (UC identity federation)
* Shareable objects: Tables, Views (including dynamic), Volumes, Models
* NOT shareable: Notebooks, Clusters, Jobs
* Recipients can read using Spark, pandas, Power BI, Tableau — not limited to Databricks
* Provider controls access; recipient never gets direct storage credentials
* Change Data Feed (CDF) can be shared for incremental consumption

**Syntax:**
```sql
-- Create a share
CREATE SHARE sales_share;

-- Add objects to a share
ALTER SHARE sales_share ADD TABLE catalog.schema.orders;
ALTER SHARE sales_share ADD VOLUME catalog.schema.my_volume;

-- Create an open recipient
CREATE RECIPIENT external_partner;

-- Grant recipient access to the share
GRANT SELECT ON SHARE sales_share TO RECIPIENT external_partner;

-- Consumer side: create a catalog from a provider share
CREATE CATALOG shared_data FROM PROVIDER my_provider SHARE sales_share;
```

### 2. Functions (UDFs)

**Key Points:**
* **SQL UDF** — scalar expression, no side effects, used in SELECT/WHERE/etc.
* **SQL Table Function** — returns a TABLE, invoked in FROM clause
* **Python UDF** — registered in UC, callable from any SQL context
* UDFs are stored in Unity Catalog (three-level namespace: `catalog.schema.function`)
* UDFs support `DETERMINISTIC` / `NOT DETERMINISTIC` hints
* UDFs cannot perform DDL/DML (no INSERT, CREATE, etc.)

**Syntax:**
```sql
-- SQL Scalar UDF
CREATE OR REPLACE FUNCTION catalog.schema.to_upper(s STRING)
RETURNS STRING
RETURN upper(s);

-- SQL UDF with multiple expressions (compound body)
CREATE FUNCTION catalog.schema.discount_price(price DOUBLE, pct DOUBLE)
RETURNS DOUBLE
RETURN price * (1 - pct / 100);

-- SQL Table Function
CREATE FUNCTION catalog.schema.high_value_orders(threshold DOUBLE)
RETURNS TABLE(order_id INT, amount DOUBLE)
RETURN SELECT order_id, amount FROM orders WHERE amount > threshold;

-- Python UDF in Unity Catalog
CREATE FUNCTION catalog.schema.mask_email(email STRING)
RETURNS STRING
LANGUAGE PYTHON
AS $$
  parts = email.split('@')
  return parts[0][:2] + '***@' + parts[1]
$$;

-- Usage
SELECT catalog.schema.to_upper(name) FROM customers;
SELECT * FROM catalog.schema.high_value_orders(1000);
```

### 3. Stored Procedures

**Key Points:**
* Can execute DDL/DML (CREATE TABLE, INSERT, MERGE, etc.) — unlike UDFs
* Support multi-statement logic with variables, IF/ELSE, loops
* Languages: **SQL** and **Python**
* Security: **INVOKER** (default — runs as caller) or **DEFINER** (runs as owner)
* Called with `CALL`, not in SELECT
* Can return a result set (last SELECT in body) or nothing
* Registered in Unity Catalog (three-level namespace)

**Syntax:**
```sql
-- SQL Procedure
CREATE OR REPLACE PROCEDURE catalog.schema.archive_old_orders(cutoff_date DATE)
SQL SECURITY INVOKER
BEGIN
  INSERT INTO catalog.schema.orders_archive
    SELECT * FROM catalog.schema.orders WHERE order_date < cutoff_date;
  DELETE FROM catalog.schema.orders WHERE order_date < cutoff_date;
END;

-- SQL Procedure with variables and control flow
CREATE PROCEDURE catalog.schema.conditional_load(src STRING)
BEGIN
  DECLARE row_count INT;
  SET row_count = (SELECT COUNT(*) FROM IDENTIFIER(src));
  IF row_count > 0 THEN
    INSERT INTO catalog.schema.target SELECT * FROM IDENTIFIER(src);
  END IF;
END;

-- Python Procedure
CREATE PROCEDURE catalog.schema.compute_stats(table_name STRING)
LANGUAGE PYTHON
AS $$
  df = spark.table(table_name)
  stats = df.summary()
  stats.write.mode("overwrite").saveAsTable("catalog.schema.stats_output")
$$;

-- Calling a procedure
CALL catalog.schema.archive_old_orders('2025-01-01');
```

### 6. Views

**Key Points:**
* **Standard View** — logical query, no storage, always reads fresh data, persisted in UC
* **Temporary View** — session-scoped, NOT in UC, lost on session end
* **Global Temporary View** — visible across sessions on same cluster (stored in `global_temp` db)
* **Dynamic View** — uses `current_user()`, `is_account_group_member()` for row/column-level security
* Views reference underlying tables; if table is dropped, view remains but queries fail
* Views can be shared via Delta Sharing (including dynamic views for fine-grained access)
* Views do NOT store data — cost zero storage

**Syntax:**
```sql
-- Standard View (persisted in UC)
CREATE OR REPLACE VIEW catalog.schema.active_customers AS
SELECT * FROM catalog.schema.customers WHERE status = 'active';

-- Temporary View (session-scoped)
CREATE OR REPLACE TEMP VIEW tmp_recent_orders AS
SELECT * FROM orders WHERE order_date > current_date() - INTERVAL 7 DAYS;

-- Global Temporary View
CREATE OR REPLACE GLOBAL TEMP VIEW global_summary AS
SELECT region, SUM(revenue) AS total FROM sales GROUP BY region;
-- Query: SELECT * FROM global_temp.global_summary;

-- Dynamic View (row-level security)
CREATE OR REPLACE VIEW catalog.schema.secure_orders AS
SELECT * FROM catalog.schema.orders
WHERE
  CASE
    WHEN is_account_group_member('admin_group') THEN TRUE
    ELSE owner_email = current_user()
  END;

-- Dynamic View (column masking)
CREATE OR REPLACE VIEW catalog.schema.masked_customers AS
SELECT
  customer_id,
  CASE
    WHEN is_account_group_member('pii_access') THEN email
    ELSE '***MASKED***'
  END AS email
FROM catalog.schema.customers;
```

### 7. Materialized Views (MV)

**Key Points:**
* Physically stores precomputed results (unlike standard views)
* Supports **incremental refresh** — only recomputes changed data when possible
* Refresh modes: **SCHEDULE** (cron/interval), **ON UPDATE** (source change), or **manual** (`REFRESH`)
* Queries return stored (possibly stale) data instantly — never block for refresh
* Created in Databricks SQL or within Spark Declarative Pipelines
* MVs are Delta tables under the hood — support time travel, VACUUM, etc.
* Cannot INSERT/UPDATE/DELETE into an MV directly
* Minimum schedule interval: 1 minute (not seconds)
* Great for: expensive aggregations, dashboards, repeated query patterns

**Syntax:**
```sql
-- Basic Materialized View
CREATE OR REPLACE MATERIALIZED VIEW catalog.schema.daily_revenue AS
SELECT date, region, SUM(amount) AS total_revenue
FROM catalog.schema.orders
GROUP BY date, region;

-- MV with scheduled refresh (every hour)
CREATE OR REPLACE MATERIALIZED VIEW catalog.schema.hourly_stats
SCHEDULE EVERY 1 HOUR AS
SELECT hour(ts) AS hr, COUNT(*) AS event_count
FROM catalog.schema.events
GROUP BY hour(ts);

-- MV with CRON schedule
CREATE OR REPLACE MATERIALIZED VIEW catalog.schema.nightly_agg
SCHEDULE CRON '0 0 2 * * ? *' AS  -- 2 AM daily
SELECT product_id, SUM(qty) AS total_qty
FROM catalog.schema.sales
GROUP BY product_id;

-- Manual refresh
REFRESH MATERIALIZED VIEW catalog.schema.daily_revenue;

-- Drop MV
DROP MATERIALIZED VIEW IF EXISTS catalog.schema.daily_revenue;
```

### 2. Auto Loader (cloudFiles)

**Key Points:**
* Incrementally ingests new files as they arrive in cloud storage
* Source format: `cloudFiles` (not `csv`, `json`, etc.)
* Two modes: **Directory Listing** (default) and **File Notification** (event-based, more scalable)
* Supports: CSV, JSON, Parquet, Avro, ORC, Text, Binary
* Schema inference + **schema evolution** (auto-detects new columns)
* `_rescued_data` column captures data that doesn't match schema
* Uses a **checkpoint** to track processed files — exactly-once guarantee
* `schemaLocation` stores inferred/evolved schema
* Option `cloudFiles.inferColumnTypes` = true → infers actual types (not just STRING)
* Option `cloudFiles.schemaHints` → override inferred types for specific columns
* Can be used in notebooks OR inside Spark Declarative Pipelines

**Schema Options:**

| Option | Purpose |
|--------|----------|
| `cloudFiles.schemaLocation` | Directory to store inferred schema |
| `cloudFiles.inferColumnTypes` | Infer types beyond STRING (default: false) |
| `cloudFiles.schemaHints` | Override specific column types |
| `cloudFiles.schemaEvolutionMode` | `addNewColumns`, `rescue`, `failOnNewColumns`, `none` |
| `rescuedDataColumn` | Column name for schema-mismatched data (default: `_rescued_data`) |

**Syntax (Notebook - PySpark):**
```python
# Basic Auto Loader
df = (spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "json")
    .option("cloudFiles.schemaLocation", "/mnt/schema/orders")
    .option("cloudFiles.inferColumnTypes", "true")
    .load("/mnt/landing/orders/")
)

# Write to Delta table with checkpoint
(df.writeStream
    .format("delta")
    .option("checkpointLocation", "/mnt/checkpoints/orders")
    .outputMode("append")
    .trigger(availableNow=True)
    .toTable("catalog.schema.orders_bronze")
)
```

**With Schema Hints:**
```python
df = (spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option("header", "true")
    .option("cloudFiles.schemaLocation", "/mnt/schema/sales")
    .option("cloudFiles.schemaHints", "price DOUBLE, quantity INT, order_date DATE")
    .option("cloudFiles.schemaEvolutionMode", "addNewColumns")
    .load("/mnt/landing/sales/")
)
```

**Inside Spark Declarative Pipelines (SQL):**
```sql
CREATE OR REFRESH STREAMING TABLE orders_bronze
AS SELECT * FROM cloud_files(
  '/mnt/landing/orders/',
  'json',
  map('cloudFiles.inferColumnTypes', 'true')
);
```

**Inside SDP (Python):**
```python
import dlt

@dlt.table
def orders_bronze():
    return (spark.readStream
        .format("cloudFiles")
        .option("cloudFiles.format", "json")
        .option("cloudFiles.inferColumnTypes", "true")
        .load("/mnt/landing/orders/")
    )
```

**File Notification Mode:**
```python
df = (spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "parquet")
    .option("cloudFiles.useNotifications", "true")  # Event-based
    .option("cloudFiles.schemaLocation", "/mnt/schema/events")
    .load("abfss://container@storage.dfs.core.windows.net/events/")
)
```

### 3. Structured Streaming

**Key Points:**
* Micro-batch processing model (not true record-at-a-time)
* Sources: Kafka, Delta, Auto Loader (cloudFiles), Rate, Socket
* Sinks: Delta, Kafka, Console, Memory, Foreach/ForeachBatch
* **Checkpoint** = mandatory for fault tolerance + exactly-once semantics
* **Output Modes**: `append` (new rows only), `complete` (full result), `update` (changed rows)
* Trigger controls batch frequency
* **Watermark** = threshold for late data; data older than watermark is dropped

**Trigger Types:**

| Trigger | Behavior |
|---------|----------|
| Default (unset) | Micro-batch as fast as possible (next batch starts immediately after previous completes) |
| `processingTime='10 seconds'` | Fixed interval micro-batches |
| `availableNow=True` | Process all available data then STOP (batch-like, replaces `once=True`) |
| `once=True` | **(Deprecated)** Process one micro-batch then stop |
| `continuous='1 second'` | Experimental low-latency continuous processing |

**Syntax:**
```python
# Default trigger (as fast as possible)
query = (df.writeStream
    .format("delta")
    .outputMode("append")
    .option("checkpointLocation", "/checkpoints/my_stream")
    .toTable("catalog.schema.target")
)

# Processing Time trigger
query = (df.writeStream
    .format("delta")
    .trigger(processingTime='30 seconds')
    .option("checkpointLocation", "/checkpoints/my_stream")
    .toTable("catalog.schema.target")
)

# Available Now (process all pending, then stop)
query = (df.writeStream
    .format("delta")
    .trigger(availableNow=True)
    .option("checkpointLocation", "/checkpoints/my_stream")
    .toTable("catalog.schema.target")
)

# Continuous trigger (experimental)
query = (df.writeStream
    .format("console")
    .trigger(continuous='1 second')
    .start()
)
```

**Checkpoint:**
* Stores offsets, committed batch IDs, state (for aggregations)
* CANNOT be shared between different queries
* Changing query logic may require a NEW checkpoint
* Location: cloud storage (ADLS, S3, GCS) or DBFS
* If deleted → stream reprocesses from scratch

**Watermark:**
```python
# Watermark for late data handling
from pyspark.sql.functions import window, col

df_with_watermark = (df
    .withWatermark("event_time", "10 minutes")  # Tolerate 10 min late data
    .groupBy(window("event_time", "5 minutes"))
    .count()
)

# Write with watermark
(df_with_watermark.writeStream
    .format("delta")
    .outputMode("append")  # append mode + watermark = drop late data
    .option("checkpointLocation", "/checkpoints/windowed")
    .toTable("catalog.schema.windowed_counts")
)
```

**foreachBatch (custom sink logic):**
```python
def process_batch(batch_df, batch_id):
    batch_df.write.mode("append").saveAsTable("catalog.schema.target")
    # Can do MERGE, multiple writes, API calls, etc.

(df.writeStream
    .foreachBatch(process_batch)
    .option("checkpointLocation", "/checkpoints/custom")
    .trigger(processingTime='1 minute')
    .start()
)
```

### 4. Spark Declarative Pipelines (SDP / formerly DLT)

**Key Points:**
* Declarative ETL framework — define WHAT, not HOW
* Object types: **Streaming Table** (append-only, incremental), **Materialized View** (batch, recomputable), **View** (temp, not persisted)
* Built-in **data quality** via `CONSTRAINT` / `EXPECT`
* Supports Python and SQL
* Automatic dependency resolution (DAG built from table references)
* Modes: **Triggered** (run once) and **Continuous** (always-on)
* Pipeline settings define target catalog/schema, compute, channel (current/preview)

**Object Types:**

| Type | Storage | Incremental | Use Case |
|------|---------|-------------|----------|
| Streaming Table | Yes | Yes (append) | Ingestion, append-only data |
| Materialized View | Yes | Partial (recomputes) | Aggregations, joins, SCD |
| View (temporary) | No | No | Intermediate transformations |

**SQL Syntax:**
```sql
-- Streaming Table (incremental append)
CREATE OR REFRESH STREAMING TABLE orders_bronze
AS SELECT * FROM cloud_files('/mnt/landing/orders/', 'json');

-- Streaming Table with quality constraints
CREATE OR REFRESH STREAMING TABLE orders_silver(
  CONSTRAINT valid_order_id EXPECT (order_id IS NOT NULL) ON VIOLATION DROP ROW,
  CONSTRAINT valid_amount EXPECT (amount > 0) ON VIOLATION FAIL UPDATE
)
AS SELECT * FROM STREAM(LIVE.orders_bronze)
WHERE status != 'cancelled';

-- Materialized View (batch, recomputable)
CREATE OR REFRESH MATERIALIZED VIEW daily_revenue
AS SELECT date, SUM(amount) AS total
FROM LIVE.orders_silver
GROUP BY date;

-- Temporary View (not persisted)
CREATE TEMPORARY LIVE VIEW orders_cleaned
AS SELECT *, upper(region) AS region_upper
FROM LIVE.orders_bronze;
```

**Python Syntax:**
```python
import dlt
from pyspark.sql.functions import *

# Streaming Table
@dlt.table(
    comment="Raw orders from landing zone"
)
def orders_bronze():
    return (spark.readStream
        .format("cloudFiles")
        .option("cloudFiles.format", "json")
        .load("/mnt/landing/orders/")
    )

# With expectations (quality rules)
@dlt.table(
    comment="Cleaned orders"
)
@dlt.expect_or_drop("valid_id", "order_id IS NOT NULL")
@dlt.expect_or_fail("valid_amount", "amount > 0")
def orders_silver():
    return dlt.read_stream("orders_bronze").filter(col("status") != "cancelled")

# Materialized View
@dlt.table(
    comment="Daily revenue aggregation"
)
def daily_revenue():
    return dlt.read("orders_silver").groupBy("date").agg(sum("amount").alias("total"))

# Temporary View
@dlt.view
def orders_temp():
    return dlt.read("orders_bronze").select("*", upper(col("region")).alias("region_upper"))
```

**Expectations (Data Quality):**

| Action | SQL Syntax | Python Decorator | Behavior |
|--------|-----------|------------------|----------|
| Warn (log) | `EXPECT (expr)` | `@dlt.expect("name", "expr")` | Log violation, keep row |
| Drop row | `EXPECT (expr) ON VIOLATION DROP ROW` | `@dlt.expect_or_drop("name", "expr")` | Remove failing rows |
| Fail pipeline | `EXPECT (expr) ON VIOLATION FAIL UPDATE` | `@dlt.expect_or_fail("name", "expr")` | Abort entire update |

**Multiple Expectations:**
```python
@dlt.expect_all_or_drop({
    "valid_id": "order_id IS NOT NULL",
    "valid_amount": "amount > 0",
    "valid_date": "order_date <= current_date()"
})
```

**Reading from other tables in pipeline:**
```python
# Streaming read (for streaming tables)
dlt.read_stream("source_table")

# Batch read (for materialized views)
dlt.read("source_table")
```

```sql
-- SQL streaming read
SELECT * FROM STREAM(LIVE.source_table)

-- SQL batch read
SELECT * FROM LIVE.source_table
```

### 5. Deletion Vectors

**Key Points:**
* Performance optimization for DELETE/UPDATE/MERGE on Delta tables
* Instead of rewriting entire Parquet files, marks deleted rows in a **separate bitmap file**
* Reads skip marked rows at scan time — faster writes, same read correctness
* Enabled by default on new tables (DBR 14.0+)
* `REORG TABLE ... APPLY (PURGE)` physically removes deletion vectors (rewrites files)
* Compatible with Photon, Z-Order, Liquid Clustering
* Reduces **write amplification** significantly for MERGE/UPDATE/DELETE

**Syntax:**
```sql
-- Enable deletion vectors on existing table
ALTER TABLE catalog.schema.my_table
SET TBLPROPERTIES ('delta.enableDeletionVectors' = true);

-- Disable deletion vectors
ALTER TABLE catalog.schema.my_table
SET TBLPROPERTIES ('delta.enableDeletionVectors' = false);

-- Purge deletion vectors (physically rewrite files)
REORG TABLE catalog.schema.my_table APPLY (PURGE);

-- Check table properties
SHOW TBLPROPERTIES catalog.schema.my_table;
DESCRIBE DETAIL catalog.schema.my_table;
```

### 6. Table Optimization & Data Layout

#### OPTIMIZE
* Compacts small files into larger files (target ∼1 GB)
* Reduces file count → faster reads
* Idempotent — safe to run repeatedly
* Does NOT delete old files (use VACUUM for that)
* Can target specific partitions

```sql
OPTIMIZE catalog.schema.orders;
OPTIMIZE catalog.schema.orders WHERE date >= '2026-01-01';
OPTIMIZE catalog.schema.orders ZORDER BY (customer_id, order_date);
```

#### Z-ORDER
* Co-locates related data in same files by column values
* Best for columns in WHERE / JOIN filters
* Applied during OPTIMIZE (not standalone)
* **NOT compatible** with Liquid Clustering
* Effectiveness decreases with more columns (max 3-4 practical)

#### Partitioning
* Physical directory-based layout
* Best for **low-cardinality** columns (date, region)
* Over-partitioning = small file problem
* Cannot change partition columns after creation
* Being replaced by Liquid Clustering

```sql
CREATE TABLE catalog.schema.orders (
  order_id BIGINT, amount DOUBLE, order_date DATE
) PARTITIONED BY (order_date);
```

#### Liquid Clustering
* **Replaces** both partitioning + Z-Order
* Clustering columns can be **changed** without full rewrite
* Incremental — only recluster newly written data on OPTIMIZE
* Up to 4 clustering columns
* NOT compatible with partitioning or Z-Order on same table
* Recommended for **all new tables**

```sql
-- Create with Liquid Clustering
CREATE TABLE catalog.schema.orders (
  order_id BIGINT, customer_id BIGINT, amount DOUBLE, order_date DATE
) CLUSTER BY (customer_id, order_date);

-- Change clustering columns (no rewrite!)
ALTER TABLE catalog.schema.orders CLUSTER BY (region, order_date);

-- Remove clustering
ALTER TABLE catalog.schema.orders CLUSTER BY NONE;

-- Trigger clustering
OPTIMIZE catalog.schema.orders;
```

#### Auto Liquid Clustering
* Databricks automatically selects optimal clustering columns
* Based on query patterns and workload analysis
* Part of Predictive Optimization
* No manual `CLUSTER BY` needed

#### ANALYZE TABLE (Column Statistics)
* Collects stats for Cost-Based Optimizer (CBO)
* Stats: row count, size, min/max, null count, distinct count
* Delta auto-collects file-level min/max for first **32 columns**
* Config: `delta.dataSkippingNumIndexedCols`

```sql
ANALYZE TABLE catalog.schema.orders COMPUTE STATISTICS;
ANALYZE TABLE catalog.schema.orders COMPUTE STATISTICS FOR COLUMNS customer_id, amount;

-- Change indexed columns for data skipping
ALTER TABLE catalog.schema.orders
SET TBLPROPERTIES ('delta.dataSkippingNumIndexedCols' = 50);
```

#### Predictive Optimization
* Auto-runs OPTIMIZE and VACUUM based on workload
* Enabled at catalog/schema/table level
* Eliminates manual scheduling
* Handles Auto Liquid Clustering column selection
* Requires UC managed tables

```sql
ALTER SCHEMA catalog.schema
SET PROPERTIES ('delta.enablePredictiveOptimization' = 'true');
```

#### Auto Optimize (Optimized Writes + Auto Compaction)

**Key Points:**
* **Optimized Writes** = coalesces small writes into optimally-sized files (~128 MB) during write
* **Auto Compaction** = automatically compacts small files AFTER a write completes (async, best-effort)
* Both are **enabled by default** on Unity Catalog managed tables (DBR 13.3+)
* Target file sizes:
  * OPTIMIZE target: **1 GB** (1024 MB)
  * Optimized Writes target: **128 MB**
  * Auto Compaction target: **128 MB** (compacts files < 128 MB)
* Auto Compaction does NOT replace OPTIMIZE — it only handles post-write small files
* Auto Compaction runs on the same cluster after the write job completes
* Both can be enabled/disabled at table level via TBLPROPERTIES

**Default File Size Targets:**

| Operation | Target File Size | Notes |
|-----------|-----------------|-------|
| OPTIMIZE | **1 GB** (1024 MB) | Full bin-packing |
| Optimized Writes | **128 MB** | During write |
| Auto Compaction | **128 MB** | Post-write compaction |
| Delta default (no opt) | Varies (often many small files) | Depends on partition/write pattern |

**Syntax:**
```sql
-- Enable Optimized Writes
ALTER TABLE catalog.schema.orders
SET TBLPROPERTIES ('delta.autoOptimize.optimizeWrite' = true);

-- Enable Auto Compaction
ALTER TABLE catalog.schema.orders
SET TBLPROPERTIES ('delta.autoOptimize.autoCompact' = true);

-- Disable both
ALTER TABLE catalog.schema.orders
SET TBLPROPERTIES (
  'delta.autoOptimize.optimizeWrite' = false,
  'delta.autoOptimize.autoCompact' = false
);

-- Change target file size for OPTIMIZE (bytes)
ALTER TABLE catalog.schema.orders
SET TBLPROPERTIES ('delta.targetFileSize' = '134217728');  -- 128 MB

-- SparkSession level (applies to all writes in session)
-- spark.conf.set("spark.databricks.delta.optimizeWrite.enabled", "true")
-- spark.conf.set("spark.databricks.delta.autoCompact.enabled", "true")
```

**PySpark:**
```python
# Enable at write time
(df.write
    .format("delta")
    .option("optimizeWrite", "true")
    .option("autoCompact", "true")
    .mode("append")
    .saveAsTable("catalog.schema.orders")
)

# Session-level
spark.conf.set("spark.databricks.delta.optimizeWrite.enabled", "true")
spark.conf.set("spark.databricks.delta.autoCompact.enabled", "true")
```

**Auto Optimize vs OPTIMIZE vs Predictive Optimization:**

| Feature | Optimized Writes | Auto Compaction | OPTIMIZE (manual) | Predictive Optimization |
|---------|-----------------|-----------------|-------------------|-------------------------|
| When | During write | After write | On demand / scheduled | Automatic (background) |
| Target size | 128 MB | 128 MB | 1 GB | 1 GB |
| Z-Order/Cluster | No | No | Yes | Yes |
| Replaces OPTIMIZE? | No | No | N/A | Yes |
| Default (UC) | Enabled | Enabled | Manual | Opt-in |

| Feature | Partitioning | Z-Order | Liquid Clustering |
|---------|-------------|---------|-------------------|
| Layout | Directory-based | Within-file co-location | Flexible clustering |

#### Window Functions
```sql
-- Ranking
SELECT
  row_number() OVER (PARTITION BY dept ORDER BY salary DESC) AS rn,    -- unique, no gaps
  rank() OVER (PARTITION BY dept ORDER BY salary DESC) AS rnk,         -- ties get same rank, gaps after
  dense_rank() OVER (PARTITION BY dept ORDER BY salary DESC) AS drnk,  -- ties get same rank, NO gaps
  ntile(4) OVER (ORDER BY salary) AS quartile                          -- divide into N buckets
FROM employees;

-- Aggregate windows
SELECT
  SUM(amount) OVER (PARTITION BY customer_id ORDER BY order_date
    ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS running_total,
  AVG(amount) OVER (PARTITION BY region) AS region_avg,
  COUNT(*) OVER () AS total_count
FROM orders;

-- Offset functions
SELECT
  lag(amount, 1, 0) OVER (ORDER BY order_date) AS prev_amount,    -- previous row
  lead(amount, 1, 0) OVER (ORDER BY order_date) AS next_amount,   -- next row
  first_value(amount) OVER (PARTITION BY cust ORDER BY dt) AS first_order,
  last_value(amount) OVER (PARTITION BY cust ORDER BY dt
    ROWS BETWEEN UNBOUNDED PRECEDING AND UNBOUNDED FOLLOWING) AS last_order
FROM orders;

-- Frame specification
ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW    -- default for ORDER BY
ROWS BETWEEN 3 PRECEDING AND 3 FOLLOWING             -- sliding window
RANGE BETWEEN INTERVAL 7 DAYS PRECEDING AND CURRENT ROW  -- time-based range
```

**PySpark:**
```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number, rank, dense_rank, lag, lead, sum as _sum

w = Window.partitionBy("dept").orderBy(col("salary").desc())

df.withColumn("rn", row_number().over(w))
df.withColumn("prev_salary", lag("salary", 1).over(w))
df.withColumn("running_total", _sum("amount").over(
    Window.partitionBy("customer_id").orderBy("order_date").rowsBetween(Window.unboundedPreceding, 0)
))
```

#### Watermark (Streaming)
```python
# Define watermark on event time column
df_watermarked = df.withWatermark("event_time", "10 minutes")

# Use with windowed aggregation
from pyspark.sql.functions import window

result = (df_watermarked
    .groupBy(window("event_time", "5 minutes"))  # 5-min tumbling window
    .agg(count("*").alias("event_count"))
)

# Sliding window
result = (df_watermarked
    .groupBy(window("event_time", "10 minutes", "5 minutes"))  # 10-min window, 5-min slide
    .agg(sum("amount").alias("total"))
)
```

**Key Rules:**
* Watermark must be defined BEFORE the aggregation
* With `append` output mode: results emitted only after watermark passes the window end
* With `update` output mode: results emitted immediately, updated in-place
* Watermark column must be a timestamp type
* State for windows older than watermark is dropped (memory management)

### 11. Column Masks

**Key Points:**
* Apply a **masking function** to a column — different users see different values based on identity/group
* Implemented via a **SQL UDF** that returns the masked value
* Applied at the **table level** in Unity Catalog (not on views)
* Mask is enforced regardless of access path (SQL, Spark, BI tools)
* The masking function receives the column value + optionally other columns as context
* Only the table OWNER or users with `MANAGE` privilege can set/remove masks
* Masking functions must be registered in Unity Catalog
* Column masks are evaluated **at query time** — no data is physically changed
* Users with the unmasked column permission (`UNMASK`) see raw values

**Syntax:**
```sql
-- Step 1: Create the masking function
CREATE OR REPLACE FUNCTION catalog.schema.mask_ssn(ssn STRING)
RETURNS STRING
RETURN
  CASE
    WHEN is_account_group_member('pii_admins') THEN ssn
    ELSE CONCAT('***-**-', RIGHT(ssn, 4))
  END;

-- Step 2: Apply the mask to a column
ALTER TABLE catalog.schema.employees
ALTER COLUMN ssn SET MASK catalog.schema.mask_ssn;

-- Remove the mask
ALTER TABLE catalog.schema.employees
ALTER COLUMN ssn DROP MASK;

-- Masking function with additional context column
CREATE FUNCTION catalog.schema.mask_salary(salary DOUBLE, department STRING)
RETURNS DOUBLE
RETURN
  CASE
    WHEN is_account_group_member('hr_team') THEN salary
    WHEN is_account_group_member('managers') AND department = current_user_department() THEN salary
    ELSE NULL
  END;

-- Apply with column reference
ALTER TABLE catalog.schema.employees
ALTER COLUMN salary SET MASK catalog.schema.mask_salary USING COLUMNS (department);
```

**Masking at CREATE TABLE time:**
```sql
CREATE TABLE catalog.schema.customers (
  customer_id BIGINT,
  email STRING MASK catalog.schema.mask_email,
  phone STRING MASK catalog.schema.mask_phone,
  name STRING
);
```

### 12. Row Filters

**Key Points:**
* Restrict which **rows** a user can see based on identity/group
* Implemented via a **SQL UDF** that returns BOOLEAN (true = row visible)
* Applied at the **table level** in Unity Catalog
* Enforced at query time across all access paths
* Only one row filter per table (but the function can have complex logic)
* The function receives table columns as parameters to evaluate visibility
* Row filters + column masks can be combined on the same table

**Syntax:**
```sql
-- Step 1: Create the row filter function
CREATE OR REPLACE FUNCTION catalog.schema.region_filter(region STRING)
RETURNS BOOLEAN
RETURN
  CASE
    WHEN is_account_group_member('global_admins') THEN TRUE
    WHEN is_account_group_member('us_team') AND region = 'US' THEN TRUE
    WHEN is_account_group_member('eu_team') AND region = 'EU' THEN TRUE
    ELSE FALSE
  END;

-- Step 2: Apply row filter to table
ALTER TABLE catalog.schema.orders
SET ROW FILTER catalog.schema.region_filter ON (region);

-- Remove row filter
ALTER TABLE catalog.schema.orders
DROP ROW FILTER;
```

**Row Filter at CREATE TABLE time:**
```sql
CREATE TABLE catalog.schema.orders (
  order_id BIGINT,
  amount DOUBLE,
  region STRING,
  owner_email STRING
)
WITH ROW FILTER catalog.schema.region_filter ON (region);
```

### 13. Dynamic Views (Alternative Masking Approach)

**Key Points:**
* Use `CASE WHEN` with identity functions inside a VIEW definition
* Simpler than column masks for basic use cases
* Can combine row-level AND column-level security in one view
* Identity functions: `current_user()`, `is_account_group_member()`, `is_member()`
* Dynamic views can be shared via Delta Sharing
* Limitation: masking is only enforced when querying through the VIEW (not the base table)

**Syntax:**
```sql
-- Column masking via dynamic view
CREATE OR REPLACE VIEW catalog.schema.v_customers AS
SELECT
  customer_id,
  name,
  CASE
    WHEN is_account_group_member('pii_access') THEN email
    ELSE CONCAT(LEFT(email, 2), '***@', SPLIT(email, '@')[1])
  END AS email,
  CASE
    WHEN is_account_group_member('finance') THEN credit_card
    ELSE CONCAT('****-****-****-', RIGHT(credit_card, 4))
  END AS credit_card,
  region
FROM catalog.schema.customers;

-- Row + Column combined in dynamic view
CREATE OR REPLACE VIEW catalog.schema.v_employee_data AS
SELECT
  employee_id,
  name,
  department,
  CASE
    WHEN is_account_group_member('hr_team') THEN salary
    ELSE NULL
  END AS salary,
  CASE
    WHEN is_account_group_member('hr_team') THEN ssn
    ELSE CONCAT('***-**-', RIGHT(ssn, 4))
  END AS ssn
FROM catalog.schema.employees
WHERE
  is_account_group_member('hr_team')
  OR department = (SELECT dept FROM catalog.schema.user_mapping WHERE user_email = current_user());
```

### 14. Comparison: Column Masks vs Row Filters vs Dynamic Views

| Feature | Column Masks | Row Filters | Dynamic Views |
|---------|-------------|-------------|---------------|
| Applied to | Table column | Table (rows) | View definition |
| Enforced on base table | **Yes** | **Yes** | No (only through view) |
| Access path agnostic | Yes | Yes | No (must query the view) |
| Implementation | SQL UDF → `SET MASK` | SQL UDF → `SET ROW FILTER` | `CASE WHEN` in view |
| Sharable via Delta Sharing | Yes (on table) | Yes (on table) | Yes (share the view) |
| Multiple per table | One per column | One per table | N/A |
| `UNMASK` privilege bypass | Yes | No | No |
| Best for | Production tables, compliance | Multi-tenant data isolation | Simple cases, ad-hoc |

### 16. Tags & Governance (ABAC)

**Key Points:**
* Unity Catalog supports **attribute-based access control (ABAC)** using tags
* Tags are key-value pairs applied to tables, columns, schemas, catalogs

## SDP (Pipeline) Materialized Views vs Databricks SQL Materialized Views

### Key Concept

Databricks has **two different contexts** where you can create Materialized Views:
1. **Inside a Spark Declarative Pipeline (SDP/DLT)** — managed by the pipeline engine

---

### Side-by-Side Comparison

| Feature | SDP (Pipeline) MV | Databricks SQL (Standalone) MV |
|---------|-------------------|-------------------------------|
| **Created with** | `CREATE OR REFRESH MATERIALIZED VIEW` in pipeline SQL, or `@dlt.table` (batch read) in Python | `CREATE MATERIALIZED VIEW` in SQL editor/warehouse |
| **Managed by** | Pipeline engine (SDP runtime) | SQL warehouse / serverless compute |
| **Refresh trigger** | Pipeline update (triggered or continuous) | Schedule (CRON/interval), manual `REFRESH`, or ON UPDATE |
| **Refresh compute** | Pipeline cluster (defined in pipeline settings) | SQL warehouse that owns the MV |
| **Incremental refresh** | Yes — pipeline tracks what changed | Yes — engine determines if incremental is possible |
| **Refresh granularity** | All MVs in the pipeline refresh together in one update | Each MV refreshes independently on its own schedule |
| **Scheduling** | Via Jobs scheduling the pipeline, or continuous mode | Built-in `SCHEDULE EVERY` / `SCHEDULE CRON` on the MV itself |
| **Dependencies** | Declared via `LIVE.table` references (auto-DAG) | Implicit (reads from source tables directly) |
| **Data quality (EXPECT)** | ✔ Supported (constraints on rows) | ✘ Not supported |
| **Location** | Published to target catalog.schema set in pipeline settings | Created directly in specified catalog.schema |
| **Drop behavior** | Dropped when removed from pipeline code + pipeline runs | `DROP MATERIALIZED VIEW` SQL statement |
| **Ownership** | Pipeline owns the table lifecycle | Creator/warehouse owns |
| **Best for** | Multi-table ETL pipelines with dependencies | Standalone aggregation tables, dashboards, isolated MVs |

### SDP (Pipeline) Materialized View

**Key Points:**
* Defined inside a pipeline (SQL or Python notebook attached to a pipeline)
* Refreshes ONLY when a **pipeline update** runs (triggered by Job schedule or manual run)
* All tables/MVs in the pipeline refresh together as a unit (DAG execution)
* Supports data quality **expectations** (`EXPECT`, `EXPECT OR DROP`, `EXPECT OR FAIL`)
* References other pipeline tables via `LIVE.table_name`
* Cannot set an independent schedule per MV — the whole pipeline runs
* Pipeline handles compute, retries, error handling
* Ideal for **multi-hop medallion architectures** (bronze → silver → gold)

**Syntax (SQL in SDP):**
```sql
-- Materialized View inside a pipeline
CREATE OR REFRESH MATERIALIZED VIEW gold_daily_revenue
COMMENT 'Daily revenue aggregation'
AS
SELECT
  order_date,
  region,
  SUM(amount) AS total_revenue,
  COUNT(*) AS order_count
FROM LIVE.silver_orders   -- references another pipeline table
GROUP BY order_date, region;
```

**Syntax (Python in SDP):**
```python
import dlt
from pyspark.sql.functions import sum, count

@dlt.table(
    comment="Daily revenue aggregation"
)
def gold_daily_revenue():
    return (
        dlt.read("silver_orders")  # batch read = MV behavior
        .groupBy("order_date", "region")
        .agg(
            sum("amount").alias("total_revenue"),
            count("*").alias("order_count")
        )
    )
```

**Refresh Process (SDP):**
```
1. Pipeline update is triggered (manually, by Job schedule, or continuous)
2. Engine builds dependency DAG from all tables/MVs in the pipeline
3. For each MV: checks if source tables have new data since last update
4. If changes detected: recomputes (full or incremental depending on query)
5. Results written to target Delta table in catalog.schema
6. All MVs in pipeline refresh in one atomic update
```

### Databricks SQL (Standalone) Materialized View

**Key Points:**
* Created directly in a SQL warehouse (no pipeline needed)
* Each MV has its **own independent refresh schedule**
* Can be refreshed via CRON, fixed interval, ON UPDATE (source change), or manual
* Runs on the SQL warehouse that created it (or serverless)
* No dependency DAG — each MV is independent
* No data quality expectations (use a separate check or view)
* Queries always return the **last-refreshed** (possibly stale) results instantly
* Cannot reference `LIVE.` tables — directly references catalog.schema.table
* Ideal for **standalone aggregations** powering dashboards or repeated queries

**Syntax:**
```sql
-- Basic standalone MV
CREATE OR REPLACE MATERIALIZED VIEW catalog.schema.daily_revenue
AS
SELECT
  order_date,
  region,
  SUM(amount) AS total_revenue,
  COUNT(*) AS order_count
FROM catalog.schema.orders
GROUP BY order_date, region;

-- MV with hourly auto-refresh
CREATE OR REPLACE MATERIALIZED VIEW catalog.schema.hourly_stats
SCHEDULE EVERY 1 HOUR
AS
SELECT hour(event_ts) AS hr, COUNT(*) AS cnt
FROM catalog.schema.events
GROUP BY hour(event_ts);

-- MV with CRON schedule
CREATE OR REPLACE MATERIALIZED VIEW catalog.schema.nightly_summary
SCHEDULE CRON '0 0 2 * * ? *'  -- 2 AM daily
AS
SELECT product_id, SUM(qty) AS total_sold
FROM catalog.schema.sales
GROUP BY product_id;

-- MV that refreshes when source updates
CREATE OR REPLACE MATERIALIZED VIEW catalog.schema.latest_agg
SCHEDULE ON UPDATE
AS
SELECT category, AVG(price) AS avg_price
FROM catalog.schema.products
GROUP BY category;

-- Manual refresh
REFRESH MATERIALIZED VIEW catalog.schema.daily_revenue;

-- Drop
DROP MATERIALIZED VIEW catalog.schema.daily_revenue;
```

**Refresh Process (DBSQL):**
```
1. Refresh triggered by: schedule (CRON/interval), ON UPDATE event, or manual REFRESH command
2. SQL warehouse (or serverless) picks up the refresh job
3. Engine checks Delta log of source tables for changes since last refresh
4. If incremental refresh is possible (simple filters/aggregations): only process new data
5. If not (complex joins, non-deterministic functions): full recomputation
6. Results atomically written to the MV's underlying Delta table
7. Queries continue reading stale data during refresh; new data visible after commit
```

### Refresh Strategies Compared

| Refresh Type | DBSQL Syntax | SDP Equivalent | Behavior |
|-------------|-------------|----------------|----------|
| Fixed interval | `SCHEDULE EVERY 1 HOUR` | Job scheduling pipeline every 1 hour | Predictable cadence |
| CRON | `SCHEDULE CRON '0 0 8 * * ?'` | Job with CRON trigger on pipeline | Time-of-day control |
| On source change | `SCHEDULE ON UPDATE` | Continuous pipeline mode | Reacts to new data |
| Manual | `REFRESH MATERIALIZED VIEW mv` | Manual pipeline update trigger | On-demand |
| Batch (process + stop) | N/A | Job with `availableNow` trigger on pipeline | Incremental batch |

### Incremental Refresh — When Does It Work?

**Conditions for incremental refresh (both SDP and DBSQL):**
* Source is a Delta table with append-only changes (or CDF enabled)
* Query uses supported operations: `SELECT`, `WHERE`, `GROUP BY`, simple `JOIN`
* No non-deterministic functions (`rand()`, `current_timestamp()`, etc.)
* No `LIMIT`, `DISTINCT` on streaming source, or complex subqueries

**Falls back to FULL refresh when:**
* Source data was updated/deleted (not just appended)
* Query uses unsupported patterns (outer joins on streaming, non-deterministic UDFs)
* Schema changed in source table
* First-ever refresh (always full)

### Common Pitfall: `LIVE.table` vs Direct Table Reference

```sql
-- INSIDE SDP pipeline: use LIVE. prefix
CREATE OR REFRESH MATERIALIZED VIEW gold_summary
AS SELECT * FROM LIVE.silver_orders;  -- ✔ pipeline-managed reference

-- OUTSIDE pipeline (DBSQL): use fully qualified name
CREATE MATERIALIZED VIEW catalog.schema.gold_summary
AS SELECT * FROM catalog.schema.silver_orders;  -- ✔ direct UC reference

-- WRONG: using LIVE. outside a pipeline
CREATE MATERIALIZED VIEW catalog.schema.summary
AS SELECT * FROM LIVE.orders;  -- ✘ ERROR: LIVE. only works inside SDP
```

**Q76.** A Materialized View inside an SDP pipeline refreshes when:  
A) It has its own CRON schedule  
B) A pipeline update is triggered (all pipeline objects refresh together)  
C) The SQL warehouse auto-detects changes  
D) Every time a user queries it  

<details><summary>Answer</summary>B — Pipeline MVs refresh when the pipeline update runs. They don't have independent schedules; the whole pipeline executes as a unit.</details>

**Q77.** A standalone DBSQL Materialized View with `SCHEDULE EVERY 1 HOUR` — what happens when a user queries it 30 minutes after the last refresh?  
A) The query triggers an immediate refresh  
B) The query returns the last-refreshed (30-minute stale) data instantly  
C) The query blocks until the next scheduled refresh  
D) The query fails with a staleness error  

<details><summary>Answer</summary>B — DBSQL MVs always return stored results immediately. Refresh happens independently on schedule.</details>

**Q78.** Which feature is available in SDP Materialized Views but NOT in DBSQL Materialized Views?  
A) Incremental refresh  
B) Data quality expectations (EXPECT / CONSTRAINT)  
C) CRON-based scheduling  
D) Delta table storage  

<details><summary>Answer</summary>B — Data quality constraints (`EXPECT`, `ON VIOLATION DROP ROW`, etc.) are only available inside SDP pipelines.</details>

**Q79.** In SDP SQL, `LIVE.silver_orders` means:  
A) A reference to a table in the current pipeline's DAG  
B) A reference to a table that is always real-time  
C) A temporary view that exists only during the query  
D) An external table from another workspace  

<details><summary>Answer</summary>A — `LIVE.table_name` references another dataset defined within the same SDP pipeline, enabling automatic dependency tracking.</details>

**Q80.** A DBSQL MV uses `SCHEDULE ON UPDATE`. When does it refresh?  
A) Immediately when any table in the workspace changes  
B) When the specific source tables referenced in the MV definition have new data  
C) Every minute regardless of changes  
D) Only when manually triggered  

<details><summary>Answer</summary>B — `ON UPDATE` monitors the source tables in the MV definition and triggers a refresh when they receive new data.</details>

**Q81.** An engineer wants to create a gold-layer aggregation that depends on a silver-layer streaming table, with data quality checks. Which approach is correct?  
A) Create a DBSQL MV with SCHEDULE EVERY 1 HOUR  
B) Create an MV inside an SDP pipeline with expectations and `LIVE.silver_table` reference  
C) Create a standard VIEW  
D) Use a stored procedure with a MERGE statement  

<details><summary>Answer</summary>B — SDP supports pipeline dependencies (LIVE. references), streaming tables, and data quality expectations — ideal for medallion architecture.</details>

**Q82.** When does a Materialized View (either SDP or DBSQL) fall back to a FULL refresh instead of incremental?  
A) Always on the first refresh  
B) When source data has UPDATEs or DELETEs (not just appends)  
C) When the query uses non-deterministic functions  
D) All of the above  

<details><summary>Answer</summary>D — Full refresh occurs on first run, when source data is modified (not append-only), or when the query pattern doesn't support incremental.</details>

**Q83.** Which compute runs the refresh for a DBSQL Materialized View?  
A) The pipeline cluster  
B) The SQL warehouse that created the MV (or serverless)  
C) The user's interactive cluster  
D) A dedicated refresh cluster  

<details><summary>Answer</summary>B — DBSQL MVs are refreshed by the SQL warehouse (or serverless compute) that owns them.</details>

**Q84.** Can you set a per-MV refresh schedule inside an SDP pipeline?  
A) Yes — each MV has its own CRON inside the pipeline  
B) No — all objects in the pipeline refresh together during a pipeline update  
C) Yes — but only for Materialized Views, not Streaming Tables  
D) Yes — using the `REFRESH INTERVAL` property  

<details><summary>Answer</summary>B — In SDP, all pipeline objects (streaming tables, MVs, views) refresh together. You cannot set independent per-MV schedules within a single pipeline.</details>

**Q85.** To give a single DBSQL MV a completely independent refresh that doesn't affect other tables, you should:  
A) Put it inside a pipeline with only that one MV  
B) Create it as a standalone DBSQL MV with its own `SCHEDULE`  
C) Both A and B work, but B is simpler  
D) Use a TEMPORARY VIEW instead  

<details><summary>Answer</summary>C — Both work, but a standalone DBSQL MV with its own schedule is the simplest approach when you don't need pipeline features like expectations or multi-table DAGs.</details>
